# Correlations for stocks I am interested in

* Time Frame: 5Y 

```python
import yfinance as yf
import numpy as np
import pandas as pd

univ = ['RR','CRML','OPEN','DVLT','BBAI', 'BITF', 'IXHL', 'ONDS', 'RGTI', 'ASTS', 'LPTH', 'BURU', 'PATH', 'DNUT', 'GLD', 'DEFT']
px = yf.download(univ, start="2020-01-01", auto_adjust=False)['Adj Close']
ret = px / px.shift() - 1
```

```python
ret.corr()
```
